```python
a = 5
b = 3

print(f"Suma: {a + b}")
print(f"Resta: {a - b}")
print(f"Multiplicación: {a * b}")
print(f"División: {a / b}")
print(f"División entera: {a // b}")
print(f"Módulo: {a % b}")
print(f"Exponenciación: {a ** b}")
```

```python
number1 = float(input("Introduce el primer número: "))
number2 = float(input("Introduce el segundo número: "))

addition = number1 + number2
subtraction = number1 - number2
multiplication = number1 * number2
division = number1 / number2 if number2 != 0 else "No se puede dividir por cero"

print("\n--- Resultados de la calculadora ----")
print(f"Suma: {number1} + {number2} = {addition}")
print(f"Resta: {number1} - {number2} = {subtraction}")
print(f"Multiplicación: {number1} x {number2} = {multiplication}")
print(f"División: {number1} / {number2} = {division}")
```
