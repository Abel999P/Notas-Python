```python
# Comparación de numeros
# obtener la entrada de los dos numeros
num1 = float(input("Ingrese el primer numero"))
num2 = float(input("Ingrese el segundo numero"))

# Comparar numeros
if num1 == 0 or num2 == 0:
    print("No se pueden comparar numeros con 0")
elif num1 > num2:
    print(f"{num1} es mayor que {num2}")
elif num1 < num2:
    print(f"{num1} es menor que {num2}")
elif num1 == num2:
    print(f"{num1} es igual que {num2}")
```
