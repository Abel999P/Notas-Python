```python
mi_lista = ["Leche","Huevos","Azucar"]

print(mi_lista)
mi_lista.append("Pan")
mi_lista.insert(1,"Mantequilla")
print(mi_lista)
mi_lista.remove("Azucar")
print(mi_lista)
mi_lista.pop()
print(mi_lista)
mi_lista.sort()
print(mi_lista)
```

```python
mi_lista.clear()
def mi_menu():
    print("\n-----Menu de lista de compras-----")
    print("1. Agregar un item")
    print("2. Eliminar un item")
    print("3. Mostrar la lista")
    print("4. Limpiar la lista")
    print("5. Salir")
    opcion = input("Seleccione una opción: ")
    return opcion

while True:
    opcion = mi_menu()
    if opcion == "1":
        item = input("Ingrese el item a agregar: ")
        mi_lista.append(item)
        print(f"{item} ha sido agregado a la lista.")
    elif opcion == "2":
        item = input("Ingrese el item a eliminar: ")
        if item in mi_lista:
            mi_lista.remove(item)
            print(f"{item} ha sido eliminado de la lista.")
        else:
            print(f"{item} no se encuentra en la lista.")
    elif opcion == "3":
        print("Lista de compras:")
        for index, item in enumerate(mi_lista):
            print(f"{index}: {item}")
    elif opcion == "4":
        mi_lista.clear()
        print("La lista ha sido limpiada.")
    elif opcion == "5":
        print("Saliendo del programa. ¡Hasta luego!")
        break
    else:
        print("Opción no válida. Por favor, seleccione una opción del 1 al 5.")
```
