```python
with open ("notas.txt","w") as file:
    file.write("")

FILE_NAME = "notas.txt"

def mostrar_menu():
    print("Seleccione una opción:")
    print("1. Leer archivo")
    print("2. Escribir en archivo")
    print("3. Añadir al archivo")
    print("4. Borrar contenido del archivo")
    print("5. Salir")

def leer_archivo():
    try:
        with open(FILE_NAME, "r") as file:
            contenido = file.read()
            print("Contenido del archivo:")
            print(contenido)
    except FileNotFoundError:
        print("El archivo no existe.")
def escribir_archivo():
    try:
        nota = input("Ingrese el texto para escribir en el archivo: ")
        with open(FILE_NAME, "w") as file:
            file.write(nota+"\n")
        print("Contenido escrito en el archivo.")
    except Exception as e:
        print(f"Ocurrió un error: {e}")

def anadir_archivo():
    try:
        nota = input("Ingrese el texto para añadir al archivo: ")
        with open(FILE_NAME, "a") as file:
            file.write("\n"+nota)
        print("Contenido añadido al archivo.")
    except Exception as e:
        print(f"Ocurrió un error: {e}")
def borrar_contenido():
    try:
        with open(FILE_NAME, "w") as file:
            file.write("")
        print("Contenido del archivo borrado.")
    except Exception as e:
        print(f"Ocurrió un error: {e}")
while True:
    mostrar_menu()
    opcion = input("Ingrese el número de la opción deseada: ")
    if opcion == "1":
        leer_archivo()
    elif opcion == "2":
        escribir_archivo()
    elif opcion == "3":
        anadir_archivo()
    elif opcion == "4":
        borrar_contenido()
    elif opcion == "5":
        print("Saliendo del programa.")
        break
    else:
        print("Opción no válida. Por favor, Ingrese un un numero del 1-5.")
```
