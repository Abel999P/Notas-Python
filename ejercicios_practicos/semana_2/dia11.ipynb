```python
try:
    num = int(input("Ingrese un numero : "))
    result = 10 / num
except ZeroDivisionError as e:
    print("Error: No se puede dividir por cero.")
except ValueError as e:
    print("Error: Entrada invalida. Por favor ingrese un numero entero.")
except Exception as e:
    print(f"Ocurrio un error inesperado: {e}")
else:
    print(f"Resultado {result}")
finally:
    print("Ejecucion finalizada donde se controlo la exepcion.")
print("HOla")
```

```python
def retirar_dinero(cantidad):
    if cantidad < 0:
        raise ValueError("Cantidad negativa no permitida.")
    print(f"Retirando {cantidad} pesos.")
try:
    retirar_dinero(-100)
except ValueError as e:
    print(f"Error: {e}")
```
