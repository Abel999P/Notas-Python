```python
def aritmetica(a, b):
    suma = a + b
    resta = a - b
    multiplicacion = a * b
    division = a / b if b != 0 else 'Indefinida'
    
    return suma,resta,multiplicacion,division

suma,resta,multiplicacion,division = aritmetica(10,5)

print(f"Suma: {suma}")
print(f"Resta: {resta}")
print(f"Multiplicacion: {multiplicacion}")
print(f"Division: {division}")
```

```python
def celsius_a_fahrenheit(celsius):
    fahrenheit = (celsius * 9/5) + 32
    return fahrenheit
def celsius_a_kelvin(celsius):
    kelvin = celsius + 273.15
    return kelvin
def fahrenheit_a_celsius(fahrenheit):
    celsius = (fahrenheit - 32) * 5/9
    return celsius
def fahrenheit_a_kelvin(fahrenheit):
    kelvin = (fahrenheit - 32) * 5/9 + 273.15
    return kelvin
def kelvin_a_celsius(kelvin):
    celsius = kelvin - 273.15
    return celsius
def kelvin_a_fahrenheit(kelvin):
    fahrenheit = (kelvin - 273.15) * 9/5 + 32
    return fahrenheit

def mostrar_menu():
    print("----- Convertidor de temperatura -----")
    print("1. Celsius a Fahrenheit")
    print("2. Celsius a Kelvin")
    print("3. Fahrenheit a Celsius")
    print("4. Fahrenheit a Kelvin")
    print("5. Kelvin a Celsius")
    print("6. Kelvin a Fahrenheit")

while True:
    mostrar_menu()
    opcion = input("Ingrese el número de la conversión que desea realizar (o 'salir' para terminar): ")
    if opcion.lower() == 'salir':
        print("Programa terminado.")
        break
    elif opcion == '1':
        celsius = float(input("Ingrese la temperatura en Celsius: "))
        print(f"{celsius} °C = {celsius_a_fahrenheit(celsius)} °F")
    elif opcion == '2':
        celsius = float(input("Ingrese la temperatura en Celsius: "))
        print(f"{celsius} °C = {celsius_a_kelvin(celsius)} K")
    elif opcion == '3':
        fahrenheit = float(input("Ingrese la temperatura en Fahrenheit: "))
        print(f"{fahrenheit} °F = {fahrenheit_a_celsius(fahrenheit)} °C")
    elif opcion == '4':
        fahrenheit = float(input("Ingrese la temperatura en Fahrenheit: "))
        print(f"{fahrenheit} °F = {fahrenheit_a_kelvin(fahrenheit)} K")
    elif opcion == '5':
        kelvin = float(input("Ingrese la temperatura en Kelvin: "))
        print(f"{kelvin} K = {kelvin_a_celsius(kelvin)} °C")
    elif opcion == '6':
        kelvin = float(input("Ingrese la temperatura en Kelvin: "))
        print(f"{kelvin} K = {kelvin_a_fahrenheit(kelvin)} °F")
    else:
        print("Opción no válida. Por favor, intente nuevamente.")
    
```
