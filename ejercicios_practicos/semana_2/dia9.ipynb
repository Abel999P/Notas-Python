```python
set_a = {"harina","azucar","manteca"}
set_b = {"azucar","huevos"}
print(set_a | set_b) # Union
print(set_a & set_b) # Intersección
print(set_a - set_b) # Diferencia
print(set_a ^ set_b) # Diferencia simétrica
print(set_a <= set_b) # Subconjunto
print(set_a >= set_b) # Superconjunto
```

```python
receta_ingredientes = {"harina","azucar","manteca","huevos","leche"}

user_input = input("Ingresa los ingredientes que tienes ")
user_ingredientes = set(user_input.split(","))

ingredientes_faltantes = receta_ingredientes - user_ingredientes
extra_ingredientes = user_ingredientes - receta_ingredientes
print("\n----- Ingredientes -----")
if ingredientes_faltantes:
    print(f"Te faltan los siguientes ingredientes: {",".join(ingredientes_faltantes)}")
if extra_ingredientes:
    print("\nTienes ingredientes adicionales que no están en la receta:")
    for ingrediente in extra_ingredientes:
        print("- " + ingrediente)
else:
    print("Tienes todos los ingredientes necesarios.")
```
