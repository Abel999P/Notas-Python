```python
def cargar_recetas(nombre_archivo):
    try:
        with open(nombre_archivo,"r") as archivo:
            contenido = archivo.read()
            recetas = contenido.split("\n\n")
            recetas_dict = {}
            for receta in recetas:
                lineas = receta.split("\n")
                if len(lineas) >= 3:
                    nombre = lineas[0].replace("Receta:","").strip()
                    ingredientes = lineas[1].replace("Ingredientes:","").strip()
                    instrucciones = lineas[2].replace("Instrucciones:","").strip()
                    recetas_dict[nombre] = {"ingredientes": ingredientes, "instrucciones": instrucciones}
            return recetas_dict
    except FileNotFoundError:
        print("No se encuentra el archivo")
        return {}

def mostrar_menu():
    print("\n--- Menú de Recetas ---")
    print("1. Ver Receta por Nombre")
    print("2. Listar Todas las Recetas")
    print("3. Salir")

def ver_receta(recetas):
    nombre = input("Introduce el nombre de la receta: ").strip()
    if nombre in recetas:
        print(f"\n--- Detalles de la Receta: {nombre} ---")
        print(f"Ingredientes: {recetas[nombre]['ingredientes']}")
        print(f"Instrucciones: {recetas[nombre]['instrucciones']}")
    else:
        print("Receta no encontrada.")

def listar_recetas(recetas):
    print("\n--- Lista de Recetas ---")
    for nombre in recetas:
        print(nombre)

ARCHIVO_RECETAS = "./file_txt/recetas.txt"
RECETAS = cargar_recetas(ARCHIVO_RECETAS)
while True:
    mostrar_menu()
    opcion = input("Selecciona una opción: ").strip()
    if opcion == "1":
        ver_receta(RECETAS)
    elif opcion == "2":
        listar_recetas(RECETAS)
    elif opcion == "3":
        print("¡Hasta luego!")
        break
    else:
        print("Opción no válida. Por favor, intenta de nuevo.")
```
