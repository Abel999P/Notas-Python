```python
ARCHIVO_DIARIO = "./file_txt/diario.txt"

def entrada_diario():
    entrada = input("Escribe tu entrada de diario: ")
    with open(ARCHIVO_DIARIO, 'a') as archivo:
        archivo.write(entrada + '\n')
    print("¡Entrada de diario guardada!")

def ver_entradas():
    try:
        with open(ARCHIVO_DIARIO,'r') as archivo:
            contenido = archivo.read()
            if contenido:
                print("\n--- Tus Entradas de Diario ---")
                print(contenido)
            else:
                print("No se encontraron entradas en el diario.")
    except FileNotFoundError:
        print("No se encontró el archivo del diario. Asegúrate de escribir una entrada primero.")

def buscar_entradas():
  palabra_clave = input("Introduce una palabra clave para buscar: ").lower()
  try:
    with open(ARCHIVO_DIARIO, 'r') as file:
      contenido = file.readlines()
      encontrado = False
      print("\n--- Resultados de Búsqueda ---")
      for entrada in contenido:
        if palabra_clave in entrada.lower():
          print(entrada.strip())
          encontrado = True
      if not encontrado:
        print("No se encontraron entradas coincidentes.")
  except FileNotFoundError:
    print("No se encontró archivo de diario. ¡Añade una entrada primero!")

def menu():
    print("\n--- Registrador de Diario ---")
    print("1. Añadir una nueva entrada")
    print("2. Ver todas las entradas")
    print("3. Buscar entradas por palabra clave")
    print("4. Salir")

while True:
    menu()
    opcion = input("Selecciona una opción (1-4): ")
    if opcion == '1':
        entrada_diario()
    elif opcion == '2':
        ver_entradas()
    elif opcion == '3':
        buscar_entradas()
    elif opcion == '4':
        print("¡Hasta luego!")
        break
    else:
        print("Opción no válida. Por favor, selecciona una opción del 1 al 4.")
```
