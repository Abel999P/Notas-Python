```python
import csv

def procesar_datos_estudiantes(archivo_entrada,archivo_salida):
    try:
        with open(archivo_entrada,"r") as archivo:
            contenido = csv.DictReader(archivo)
            informes_estudiantes = []

            for fila in contenido:
                nombre = fila["Nombre"]
                matematicas = int(fila["Matematicas"])
                ciencias = int(fila["Ciencias"])
                ingles = int(fila["Inglés"])
                promedio = round(((matematicas + ciencias + ingles) / 3),2)
                estado = "Aprobado" if promedio >= 60 else "Reprobado"

                informes_estudiantes.append({
                    "Nombre": nombre,
                    "Matematicas": matematicas,
                    "Ciencias": ciencias,
                    "Inglés": ingles,
                    "Promedio": promedio,
                    "Estado": estado
                })
        with open(archivo_salida,"w",newline="") as archivo:
            escritor = csv.DictWriter(archivo,fieldnames=["Nombre","Matematicas","Ciencias","Inglés","Promedio","Estado"])
            escritor.writeheader()
            escritor.writerows(informes_estudiantes)
        print("Datos procesados y guardados exitosamente.")
    except FileNotFoundError:
        print(f"Error: El archivo {archivo_entrada} no se encontró.")
    except KeyError:
        print("Error: El formato del archivo CSV es incorrecto. Asegúrate de que las columnas sean 'Nombre', 'Matematicas', 'Ciencias' e 'Inglés'.")
    except Exception as e:
        print(f"Ocurrió un error inesperado: {e}")

procesar_datos_estudiantes("./file_csv/alumnos.csv","./file_csv/informe_estudiantes.csv")
```
