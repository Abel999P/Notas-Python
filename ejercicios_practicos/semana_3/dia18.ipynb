```python
import json
import os

ARCHIVO_TAREAS = "./file_json/mi_tareas.json"

if not os.path.exists(ARCHIVO_TAREAS):
    with open(ARCHIVO_TAREAS,"w") as archivo:
        json.dump([],archivo)

def cargar_tareas():
    with open(ARCHIVO_TAREAS,"r") as archivo:
        return json.load(archivo)

def guardar_tareas(tareas):
    with open(ARCHIVO_TAREAS,"w") as archivo:
        json.dump(tareas,archivo, indent=4)

def agregar_tarea():
    nombre_tarea = input("Ingrese el nombre de la tarea: ").strip()
    tareas = cargar_tareas()
    tareas.append({"tarea":nombre_tarea,"estado":"incompleta"})
    guardar_tareas(tareas)
    print(f"tarea {nombre_tarea} agregada correctamente! ")

def ver_tareas():
    tareas = cargar_tareas()
    if tareas:
        print("\n--- Lista de tareas ---")
        for idx,tarea in enumerate(tareas,start=1):
            print(f"{idx} : {tarea['tarea']} - {tarea['estado']}")
    else:
        print("No tienes tareas !")
    
def modificar_tarea():
    tareas = cargar_tareas()
    ver_tareas()
    try:
        index_tarea = int(input("Ingrese el numero de la tarea a modificar: "))-1
        if 0<=index_tarea < len(tareas):
            nuevo_estado = input("Ingrese el estado (completo/incompleto): ").strip()
            tareas[index_tarea]['estado'] = nuevo_estado
            guardar_tareas(tareas)
            print("Estado de la tarea guardado !")
        else:
            print("Tarea invalida o indice incorrecto")
    except ValueError:
        print(" entrada no numérica !")
    except Exception as e:
        print(f"Ocurrio un error - {e}")
    
def eliminar_tarea():
    tareas = cargar_tareas()
    ver_tareas()
    try:
        index_tarea = int(input("Ingrese el numero de la tarea a eliminar: ")) -1
        if 0<= index_tarea < len(tareas):
            del_tarea = tareas.pop(index_tarea)
            guardar_tareas(tareas)
            print(f"tarea '{del_tarea["tarea"]}' - confirma eliminación !")
        else:
            print(" índice inválido !")
    except ValueError:
        print(f"entrada no numérica")
    except Exception as e:
        print(f"Ocurrio un error - {e}")

def mostrar_menu():
    print(f"\n--- Gestor de Tareas ---")
    print(f"1. Añadir tarea")
    print(f"2. Ver todas las tareas")
    print(f"3. Actualizar estado")
    print(f"4. Eliminar tarea")
    print(f"5. Salir")

while True:
    mostrar_menu()
    opcion = input("Ingrese una opcion de (1-5): ").strip()

    if opcion=='1':
        agregar_tarea()
    elif opcion == '2':
        ver_tareas()
    elif opcion == '3':
        modificar_tarea()
    elif opcion == '4':
        eliminar_tarea()
    elif opcion == '5':
        print("Cerrando programa ...")
        break
    else:
        print("Opcion invalida ingrese una opcion del (1-5)")
```
