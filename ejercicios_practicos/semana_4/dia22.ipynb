```python
class Auto():
    def __init__(self,marca,modelo):
        self.marca = marca
        self.modelo = modelo

    def mostrar_info(self):
        print(f"Caracteristicas del auto \n--- Marca : {self.marca} ---\n--- Modelo : {self.modelo}\n")

mi_auto = Auto("Tesla","Modelo 3")
vecino_auto = Auto("Ferrary","1")

mi_auto.mostrar_info()
vecino_auto.mostrar_info()
```

```python
class Perro():
    def __init__(self,nombre,raza):
        self.nombre = nombre
        self.raza = raza
    
    def ladrar(self):
        print(f"{self.nombre} esta Ladrando")
        
bob = Perro("Bob","Golden")
maximo = Perro("Maximo","Buldox")
bob.ladrar()
maximo.ladrar()
```

<h1>Simulador de cuentas bancarias</h1>

```python
class CuentaBank:
    def __init__(self,nombre_cuenta,saldo=0):
        self.nombre_cuenta = nombre_cuenta
        self.saldo = saldo

    def depositar_dinero(self,monto):
        if monto > 0:
            self.saldo += monto
            print(f"Tu monto de ${monto}. Fue ingresado con exito !\nTu saldo actual es {self.saldo}")
        else:
            print("Monto invalildo el monto no puede ser menor a 0")
    
    def extraer_dinero(self,monto):
        if monto>0 and monto<=self.saldo:
            self.saldo -= monto
            print(f"Extraccion de {monto}, se concreto con exito !\nTu saldo actual es {self.saldo}")
        elif monto<0:
            print("Monto invalido el monto no puede ser menor a 0")
        else:
            print("No tienes saldo suficiente")
    
    def cuenta_info(self):
        print("\n--- Informacion de tu Cuenta ---")
        print(f"Usuario:{self.nombre_cuenta}")
        print(f"Saldo  :{self.saldo}")


cuentas = {}

def crear_cuenta():
    nombre = input("Ingrese el nombre de su cuenta : ").strip()
    deposito_inicial = float(input("Ingrese un saldo inicial : "))
    cuenta = CuentaBank(nombre,deposito_inicial)
    cuentas[nombre] = cuenta
    print("Cuenta creada con exito !")

def ingresar_a_cuenta():
    nombre = input("Ingrese el nombre de la cuenta : ").strip()
    if nombre in cuentas:
        cuenta = cuentas[nombre]

        while True:
            print("--- Menu de Cuenta ---")
            print("1- Depositar")
            print("2- Extraer")
            print("3- Informacion")
            print("4- Salir")
            opcion = input("Ingresa una opcion(1-4) : ")

            if opcion=="1":
                monto = float(input("Ingrese el monto a depositar : "))
                cuenta.depositar_dinero(monto)
            elif opcion=="2":
                monto = float(input("Ingrese el monto a extraer   : "))
                cuenta.extraer_dinero(monto)
            elif opcion=="3":
                cuenta.cuenta_info()
            elif opcion=="4":
                print("Cerrando secion...")
                break
            else:
                print("Ingreso invalido --- ingrese una opcion valida (1-4)")
        else:
            print("Cuenta no encontrada ")

if __name__=="__main__":

    while True:
        print("--- Simulador de Cuenta Bancaria ---")
        print("1- Crear Cuenta ")
        print("2- Ingresar a tu cuenta")
        print("3- Salir")
        opcion = input("Ingrese una opcion(1-4) : ")
        
        if opcion=="1":
            crear_cuenta()
        elif opcion=="2":
            ingresar_a_cuenta()
        elif opcion=="3":
            print("Cerrando programa bye :) ! ....")
            break
        else:
            print("Ingreso invalido --- ingrese una opcion valida (1-3)")
            
```
