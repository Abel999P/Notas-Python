```python
class CuentaBancaria:
    def __init__(self, propietario, saldo=0):
        self.propietario = propietario
        self.saldo = saldo

    def depositar(self, cantidad):
        self.saldo += cantidad
        print(f"Deposito ${cantidad}. \nNuevo saldo: ${self.saldo}")

cuenta = CuentaBancaria("Juan Pérez", 1000)
cuenta.depositar(500)
```

```python
# Metodos de clase y metodos staticos
class Utilidad():
    app_version = "1.0"
   
    @classmethod
    def ver_version(cls):
        print(f"App version {cls.app_version}")
    
    @staticmethod
    def saludo():
        print("Hola Bienvenido a la App")


Utilidad.ver_version()
Utilidad.saludo()
```

```python
# Encapsulamiento 
class Cuenta:
    def __init__(self,propietario,saldo=0):
        self.propietario = propietario
        self.__saldo = saldo 

    def depositar(self,cantidad):
        if cantidad > 0:
            self.__saldo += cantidad
            print(f"Dpositado ${cantidad}.\nNuevo saldo: ${self.__saldo}")
        else:
            print("Cantidad a depositar no valida.")

    def obtener_saldo(self):
        return self.__saldo

cuenta = Cuenta("Jhon",1000)
cuenta.depositar(500)
print(f"Saldo de la cuenta: ${cuenta.obtener_saldo()}")
```

```python
# Biblioteca

class Libro:
    def __init__(self,titulo,autor):
        self.titulo = titulo
        self.autor = autor
        self.prestado = False
    def mostrar_info(self):
        estado = "Disponible" if not self.prestado else "Prestado"
        print(f"Titulo:{self.titulo}\nAutor:{self.autor}\nEstado:{self.prestado}")

class Biblioteca:
    def __init__(self):
        self.libros = []

    def agregar_libro(self,titulo,autor):
        nuevo_libro = Libro(titulo,autor)
        self.libros.append(nuevo_libro)
        print(f"Libro '{titulo}' de {autor} agregado a la biblioteca.")

    def ver_libros(self):
        if not self.libros:
            print("No hay libros en la biblioteca.")
        else:
            print("\n--- Catálogo de la Biblioteca ---")
            for libro in self.libros:
                libro.mostrar_info()

    def prestar_libro(self,titulo):
        for libro in self.libros:
            if libro.titulo == titulo and not libro.prestado:
                print(f"El libro '{titulo}' ha sido prestado. ¡Disfruta la lectura!")
                return
        print(f"El libro '{titulo}' no está disponible para préstamo.")

    def devolver_libro(self, titulo):
        for libro in self.libros:
            if libro.titulo == titulo and libro.prestado:
                libro.prestado = False
                print(f"El libro '{titulo}' ha sido devuelto.")
                return
        print(f"El libro '{titulo}' no está en la biblioteca.")

biblioteca = Biblioteca()

while True:
    print("\n--- Sistema de Gestión de Biblioteca ---")
    print("1. Agregar Libro")
    print("2. Ver Libros")
    print("3. Prestar Libro")
    print("4. Devolver Libro")
    print("5. Salir")

    opcion = input("Elige una opción (1-5): ").strip()

    if opcion == "1":
        titulo = input("Ingresa el título del libro: ").strip()
        autor = input("Ingresa el nombre del autor: ").strip()
        biblioteca.agregar_libro(titulo, autor)
    elif opcion == "2":
        biblioteca.ver_libros()
    elif opcion == "3":
        titulo = input("Ingresa el título del libro a prestar: ").strip()
        biblioteca.prestar_libro(titulo)
    elif opcion == "4":
        titulo = input("Ingresa el título del libro a devolver: ").strip()
        biblioteca.devolver_libro(titulo)
    elif opcion == "5":
        print("Saliendo del Sistema ( Gestión de Biblioteca ). ¡Hasta luego!")
        break
    else:
        print("Opción no válida. Por favor selecciona una opción válida (1-5).")
```
