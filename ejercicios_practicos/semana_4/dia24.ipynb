```python
from os import name

class Empleado:
    def __init__(self,nombre,id_empleado,salario):
        self.nombre = nombre 
        self.id_empleado = id_empleado
        self.salario = salario
        
    def mostrar_info(self,funcionalidad="Sin definir"):
        print("\n--- Detalles del Empleado ---")
        print(f"Funcion:{funcionalidad}")
        print(f"Nombre : {self.nombre}")
        print(f"ID del Empleado : {self.id_empleado}")
        print(f"Salario : {self.salario}")

    def calcular_bono(self):
        return self.salario * 0.1

class Gerente(Empleado):
    def __init__(self,nombre,id_empleado,salario,departamento):
        super().__init__(nombre,id_empleado,salario)
        self.departamento = departamento

    def mostrar_info(self):
        super().mostrar_info(funcionalidad="Gerente")
        print(f"Departamento {self.departamento}"
              
    def calcular_bono(self):
        return self.salario * 0.2

class Desarrollador(Empleado):
    def __init__(self,nombre,id_empleado,salario,lenguaje_programacion):
        super().__init__(nombre,id_empleado,salario)
        self.lenguaje_programacion = lenguaje_programacion

    def mostrar_info(self):
        super().mostrar_info(funcionalidad="Programador")
        print(f"Lenguaje de Programación: {self.lenguaje_programacion")

    def calcular_bono(self):
        return self.salario * 0.5

empleados = []

def agregar_empleado():
    print("\n--- Elegir Tipo de Empleado ---")
    print("1- Empleado Regular")
    print("2- Gerente")
    print("3- Desarrollador")
    opcion = int(input("Elige una opcion: ").strip())

    nombre = input("Ingresa el Nombre del Empleado: ").strip()
    id_empleado = input("Ingresa el ID del Empleado: ").strip()
    salario = float(input("Ingresa el Salario del Empleado: ").strip()

    if opcion == 1:
        empleados.append(Empleado(nombre,id_empleado,salario))
    elif opcion==2:
        departamento = input("Ingresa el Departamento: ").strip()
        empleados.append(Gerente(nombre,id_empleado,salario,departamento))
    elif opcion==3:
        lenguaje_programacion = input("Ingresa el Lenguaje de Programaci{on: ").strip()
        empleados.append(Desarrollador(nombre,id_empleado,salario,lenguaje_programacion))
    else:
        print("Opcion no válida")

def mostrar_todos_los_empleados():
    print("\n--- Todos los Empleados ---")
    for empleado in empleados:
        empleado.mostrar_info()
        print(f"Bono : {empleado.calcular_bono()}")

while True:
    print("\n--- Sistema de Gestion de Empleados ---")
    print("1- Agregar Empleado")
    print("2- Mostrar Todos los Emopleados")
    print("3- Salir")
    opcion = int(input("Elige una opcion (1-3): ").strip())

    if opcion==1:
        agregar_empleado()
    elif opcion==2:
        mostrar_todos_los_empleados()
    elif opcion==3:
        print("Saliendo del programa.")
        break
    else:
        print("Opcion no valida")
```
