Factorizacion y resolucion de ecuaciones para una variable

```python
from sympy import var,Eq
from sympy.solvers import solve

# Identificar todas las variables
var('a b c d x')

# Lados isquierdo y drecho del signo igual
izquierdo = 0
derecho = a*x**2+ b*x + c

# Variable para resolver
variable = x

# Ecuacion de sympy isquierdo = derecho
eq1 = Eq(izquierdo,derecho)

# sympy resuelve para esa variable
sol = solve(eq1,variable)

for s in sol:
    print(variable," = ",s)
```
