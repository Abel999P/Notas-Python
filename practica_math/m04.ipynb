Exponentes positivos y negativos

```python
print(10**1)
print(10**2)
print(10**3)
print(10**0)
print(10**-1)
print(10**-2)
print(10**-3)
```

La entrada del usuario viene como una cadena

```python
texto = input("Introduce un número: ")
print(texto)
print(texto+texto)
print(len(texto))
```

Convierte la entrada como una variable `float` o `int`

```python
texto = input("Introduce un número: ")
num = float(texto)
print(num+4)
```

Ahora pon todo junto

```python
# Obtener entrada de cadena que incluira un punto decimal
digitos = input("Introduce un numero decimal para convertir: ")
# Obtener el numero de decimales como un entero
exponente = int(len(digitos)) -1
# Convertir la entrada a un numero float
num = float(digitos)
# Usar el exponente para obtener el numerador
numerador = int(num* 10**exponente)
# Usar el exponente para obtener el denominador
denominador = 10**exponente
# porcentaje son los primeros dos decimales
porcentaje = num*100
#Salida
print("El decimal es ",num)
print("La fracción es ",numerador,"/",denominador)
print("El porcentaje es ",porcentaje,"%")
```

```python
x = 5
y = 4*x + 3 # funcion

print(x,",",y)
print("x \t y")

for x in range(11):
    y = 4*x + 3
    print(x,"\t",y)
```

```python
def f(x):
    y = 4*x + 3
    return y

print(5,",",f(5))

for x in range(11):
    print(x,"\t",f(x))
```

```python
# Usar una matriz como entradas

import matplotlib.pyplot as plt
import numpy as np

xmin=-10
xmax=10
ymin=-10
ymax=10
punto = 2*(xmax-xmin)

x = np.linspace(xmin,xmax,punto)

fig,ax = plt.subplots()
plt.axis([xmin,xmax,ymin,ymax])
plt.plot([xmin,xmax],[0,0],"b")
plt.plot([0,0],[ymin,ymax],"b")

y=2*x+1
plt.plot(x,y,"pink")
ax.grid(True)
ax.set_xticks(np.arange(xmin,xmax,1))
ax.set_yticks(np.arange(ymin,ymax,1))
plt.show()
```

```python
# Mas formas de personalizar un grafico
import matplotlib.pyplot as plt
import numpy as np
xmin=-10
xmax=11
ymin=-10
ymax=11
puntos = 2*(xmax-xmin)

x=np.linspace(xmin,xmax,puntos)

fig,ax = plt.subplots()
plt.axis([xmin,xmax,ymin,ymax])
plt.plot([xmin,xmax],[0,0],"b")
plt.plot([0,0],[ymin,ymax],"b")

ax.set_xlabel("valores de x")
ax.set_ylabel("valores de y")
ax.set_title("Grafico")
ax.grid(True)

ax.set_xticks(np.arange(xmin,xmax,1))
ax.set_yticks(np.arange(ymin,ymax,1))

y= 2*x + 1
plt.plot(x,y,label='y=2x+1')
plt.plot([4],[6],"ro",label="punto")
plt.plot(x,4*x,label="linea mas inclinada")
plt.legend()
plt.show()
```
