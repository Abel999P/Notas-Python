```python
# Mostrar el grafico
import matplotlib.pyplot as plt
x1=2
y1=3
x2=6
y2=8

# Desarrollar la ecuacion y=mx+b
m=(y2-y1)/(x2-x1)
b=y1-m*x1
print("y = ",m,"x + ",b)

xmin=-10
xmax=10
ymin=-10
ymax=10
y3 = m*xmin + b
y4 = m*xmax + b

#
fig,ax = plt.subplots()
plt.axis([xmin,xmax,ymin,ymax])
plt.plot([xmin,xmax],[0,0],"b")
plt.plot([0,0],[ymin,ymax],"b")

# Graficar la funcion lineal como una linea roja
plt.plot([xmin,xmax],[y3,y4],"r")
plt.show()
```

```python
from sympy import symbols
from sympy.solvers import solve

x=symbols('x')
eq = 10*x-1000
sol = solve(eq,x)
print("x = ",sol)
```
