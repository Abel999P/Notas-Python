```python
# Puedes agregar este código al código de decimal a fracción
digito = input("Ingrese un número decimal para convertir:")
# Obtenga el número decimal a convertir
exponente = int(len(digito))-1
n = float(digito)
numerador = int(n * 10**exponente)
denominador = 10**exponente

factor = 1
for test_factor in range(1,denominador+1):
    if numerador%test_factor==0 and denominador%test_factor==0:
        factor = test_factor
num = int(numerador/factor)
den = int(denominador/factor)

print("El decimal es ",n)
print("La fraccion es ",num,"/",den)
```
