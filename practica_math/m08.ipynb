```python
from sympy import symbols,linsolve

x,y = symbols('x y')

eq1= 2*x + y - 1
eq2= x - 2*y + 7

sol = linsolve([eq1,eq2],(x,y))
x_sol = sol.args[0][0]
y_sol = sol.args[0][1]

print(f"{x_sol}")
print(f"{y_sol}")
```

```python
from sympy import symbols,var,linsolve,Eq
from sympy.plotting import plot
from sympy.solvers import solve

var('x y')


eq1= 2*x + y - 1
eq2= x - 2*y + 7

sol = linsolve([eq1,eq2],(x,y))
for a in range(len(sol.args)):
    x_sol = sol.args[a][0]
    y_sol = sol.args[a][1]
    print(f"Solucion = (x,{x_sol}) (y,{y_sol})")

y_eq1 = Eq(eq1,0)
y1 = solve(y_eq1,y)

y_eq2 = Eq(eq2,0)
y2 = solve(y_eq2,y)

print(f"y = {y1[0]}")
print(f"y = {y2[0]}")

x = symbols('x')
xmin = -10
xmax = 10
plot(y1[0],y2[0],(x,xmin,xmax))
```

```python
from sympy import symbols,nonlinsolve,var,Eq
from sympy.plotting import plot
from sympy.solvers import solve

var('x y') 

first = -x**2 - y + 10
second = 2*x**2 - 2*y - 4

solution = nonlinsolve([first, second], (x, y)) 
for a in range(len(solution.args)):
    x_solution = solution.args[a][0] 
    y_solution = solution.args[a][1] 
    print("Solution = (", x_solution, ",", y_solution, ")")

y_first = Eq(first,0) 
y1 = solve(y_first,y) 

y_second = Eq(second,0) 
y2 = solve(y_second,y)

print("y = ", y1[0])
print("y = ", y2[0])

x = symbols('x')
xmin = -10
xmax = 10
plot(y1[0], y2[0], (x,xmin,xmax))  
```
