```python
import matplotlib.pyplot as plt
import numpy as np

precio_1 = 2
demanda_1 = 46
precio_2 = 10
demanda_2 = 30
m = (demanda_2 - demanda_1) / (precio_2 - precio_1)
b = demanda_1 - m*precio_1

xmin = 0
xmax = 50
ymin = 0
ymax = 320

points = 10*(xmax-xmin)
x = np.linspace(xmin,xmax,points)

fig , ax = plt.subplots()
plt.axis([xmin,xmax,ymin,ymax])
plt.plot([xmin,xmax],[0,0],"b")
plt.plot([0,0],[ymin,ymax],"b")

plt.xlabel("Precio")
plt.title("Finanzas Algebra")


demanda = m*x + b
# Linea 1
#plt.ylabel("Demanda")
#plt.plot(x,demanda)

# Linea 2
reventa = x*demanda
plt.plot(x, reventa)
plt.ylabel("Reventa")

plt.show()
```
